```python
import os
import re
import cv2
import torch
import pandas as pd
import numpy as np

from PIL import Image
from facenet_pytorch import MTCNN


# =========================================================
# 1. Paths
# =========================================================

BASE_PATH = "/kaggle/input/datasets/reubensuju/celeb-df-v2"

REAL_CELEB = os.path.join(BASE_PATH, "Celeb-real")
REAL_YOUTUBE = os.path.join(BASE_PATH, "YouTube-real")
FAKE = os.path.join(BASE_PATH, "Celeb-synthesis")


# =========================================================
# 2. Find saved sampling CSV
# =========================================================

CSV_PATH = None

for root, dirs, files in os.walk("/kaggle/input"):
    for file in files:
        if file == "celebdf_top5_final.csv":
            CSV_PATH = os.path.join(root, file)
            break
    if CSV_PATH:
        break

print("CSV path:", CSV_PATH)


# =========================================================
# 3. Load CSV
# =========================================================

sampling_df = pd.read_csv(CSV_PATH)

print("Total videos:", len(sampling_df))


# =========================================================
# 4. Convert selected frame lists
# =========================================================

def extract_numbers(value):
    text = str(value)
    return re.findall(r"-?\d+(?:\.\d+)?", text)


sampling_df["selected_frames"] = sampling_df[
    "selected_frames"
].apply(
    lambda x: [int(v) for v in extract_numbers(x)]
)

sampling_df["motion_scores"] = sampling_df[
    "motion_scores"
].apply(
    lambda x: [float(v) for v in extract_numbers(x)]
)


# =========================================================
# 5. Select first video
# =========================================================

sample_row = sampling_df.iloc[0]

sample_video_name = sample_row["video"]
sample_label = sample_row["label"]
sample_frames = sample_row["selected_frames"]


if sample_label == "Real":

    celeb_path = os.path.join(
        REAL_CELEB,
        sample_video_name
    )

    youtube_path = os.path.join(
        REAL_YOUTUBE,
        sample_video_name
    )

    if os.path.exists(celeb_path):
        sample_video_path = celeb_path
    else:
        sample_video_path = youtube_path

else:

    sample_video_path = os.path.join(
        FAKE,
        sample_video_name
    )


print("Video:", sample_video_name)
print("Label:", sample_label)
print("Selected frames:", sample_frames)
print("Video exists:", os.path.exists(sample_video_path))


# =========================================================
# 6. Read selected frames
# =========================================================

def read_selected_frames(video_path, frame_indices):

    cap = cv2.VideoCapture(video_path)

    frames = {}

    for idx in frame_indices:

        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)

        ret, frame = cap.read()

        if ret:

            frame_rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            frames[idx] = frame_rgb

    cap.release()

    return frames


selected_images = read_selected_frames(
    sample_video_path,
    sample_frames
)

print("Frames successfully read:", len(selected_images))
print("Frame numbers:", list(selected_images.keys()))


# =========================================================
# 7. Initialize MTCNN on CPU for first test
# =========================================================

mtcnn_cpu = MTCNN(
    image_size=224,
    margin=20,
    min_face_size=40,
    thresholds=[0.6, 0.7, 0.7],
    factor=0.709,
    post_process=False,
    device="cpu"
)

print("MTCNN initialized successfully.")


# =========================================================
# 8. Test ONE frame only
# =========================================================

test_frame_idx = list(selected_images.keys())[0]

test_image = Image.fromarray(
    selected_images[test_frame_idx]
)

face = mtcnn_cpu(test_image)

print("Test frame:", test_frame_idx)
print("Face detected:", face is not None)

if face is not None:
    print("Face shape:", tuple(face.shape))
```

```python
import os
import cv2
import torch
import pandas as pd
import numpy as np
from PIL import Image
from tqdm import tqdm
from facenet_pytorch import MTCNN

# =========================================================
# 1. Paths
# =========================================================

OUTPUT_ROOT = "/kaggle/working/celebdf_aligned_faces"

REAL_OUT = os.path.join(OUTPUT_ROOT, "Real")
FAKE_OUT = os.path.join(OUTPUT_ROOT, "Fake")

os.makedirs(REAL_OUT, exist_ok=True)
os.makedirs(FAKE_OUT, exist_ok=True)

CHECKPOINT_PATH = "/kaggle/working/celebdf_alignment_checkpoint.csv"


# =========================================================
# 2. MTCNN on GPU
# =========================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

mtcnn = MTCNN(
    image_size=224,
    margin=20,
    min_face_size=40,
    thresholds=[0.6, 0.7, 0.7],
    factor=0.709,
    post_process=False,
    keep_all=False,
    device=device
)

print("MTCNN ready.")


# =========================================================
# 3. Resume existing checkpoint
# =========================================================

if os.path.exists(CHECKPOINT_PATH):

    checkpoint_df = pd.read_csv(CHECKPOINT_PATH)

    processed_videos = set(
        checkpoint_df["video"].astype(str)
    )

    results = checkpoint_df.to_dict("records")

    print(
        "Resuming from:",
        len(processed_videos),
        "videos"
    )

else:

    processed_videos = set()
    results = []

    print("Starting from beginning.")


# =========================================================
# 4. Read selected frames
# =========================================================

def read_selected_frames(video_path, frame_indices):

    cap = cv2.VideoCapture(video_path)

    frames = {}

    for idx in frame_indices:

        cap.set(
            cv2.CAP_PROP_POS_FRAMES,
            int(idx)
        )

        ret, frame = cap.read()

        if ret:

            frame_rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            frames[int(idx)] = frame_rgb

    cap.release()

    return frames


# =========================================================
# 5. Process all 6529 videos
# =========================================================

SAVE_EVERY = 100

for counter, row in enumerate(
    tqdm(
        sampling_df.to_dict("records"),
        desc="MTCNN + Face Alignment"
    ),
    start=1
):

    video_name = str(row["video"])
    label = str(row["label"])

    # -----------------------------------------------------
    # Resume: skip already processed videos
    # -----------------------------------------------------

    if video_name in processed_videos:
        continue


    # -----------------------------------------------------
    # Video path
    # -----------------------------------------------------

    if label == "Real":

        celeb_path = os.path.join(
            REAL_CELEB,
            video_name
        )

        youtube_path = os.path.join(
            REAL_YOUTUBE,
            video_name
        )

        if os.path.exists(celeb_path):
            video_path = celeb_path
        else:
            video_path = youtube_path

    else:

        video_path = os.path.join(
            FAKE,
            video_name
        )


    # -----------------------------------------------------
    # Check video
    # -----------------------------------------------------

    if not os.path.exists(video_path):

        results.append({
            "video": video_name,
            "label": label,
            "selected_frames": row["selected_frames"],
            "detected_frames": [],
            "num_faces": 0,
            "status": "video_not_found"
        })

        processed_videos.add(video_name)

        continue


    # -----------------------------------------------------
    # Selected frame indices
    # -----------------------------------------------------

    frame_indices = row["selected_frames"]

    if isinstance(frame_indices, str):
        frame_indices = [
            int(x)
            for x in frame_indices.strip("[]").split(",")
            if x.strip()
        ]


    # -----------------------------------------------------
    # Read selected frames
    # -----------------------------------------------------

    selected_images = read_selected_frames(
        video_path,
        frame_indices
    )


    # -----------------------------------------------------
    # Output folder
    # -----------------------------------------------------

    label_root = (
        REAL_OUT
        if label == "Real"
        else FAKE_OUT
    )

    video_id = os.path.splitext(
        video_name
    )[0]

    video_out = os.path.join(
        label_root,
        video_id
    )

    os.makedirs(
        video_out,
        exist_ok=True
    )


    # -----------------------------------------------------
    # MTCNN + Alignment
    # -----------------------------------------------------

    detected_frames = []

    for frame_idx, frame_rgb in selected_images.items():

        try:

            image = Image.fromarray(
                frame_rgb
            )

            face = mtcnn(image)

            if face is None:
                continue


            # Convert tensor → RGB image
            face_np = (
                face.cpu()
                .permute(1, 2, 0)
                .numpy()
            )

            face_np = np.clip(
                face_np,
                0,
                255
            ).astype(np.uint8)


            aligned_face = Image.fromarray(
                face_np
            )


            output_path = os.path.join(
                video_out,
                f"frame_{frame_idx:04d}.jpg"
            )

            aligned_face.save(
                output_path,
                quality=95
            )

            detected_frames.append(
                int(frame_idx)
            )

        except Exception as e:

            print(
                f"\nError in {video_name}, "
                f"frame {frame_idx}: {e}"
            )


    # -----------------------------------------------------
    # Store metadata
    # -----------------------------------------------------

    results.append({

        "video": video_name,

        "label": label,

        "selected_frames": frame_indices,

        "detected_frames": detected_frames,

        "num_faces": len(detected_frames),

        "status": "processed"

    })

    processed_videos.add(video_name)


    # -----------------------------------------------------
    # Checkpoint every 100 videos
    # -----------------------------------------------------

    if counter % SAVE_EVERY == 0:

        checkpoint_df = pd.DataFrame(
            results
        )

        checkpoint_df.to_csv(
            CHECKPOINT_PATH,
            index=False
        )

        print(
            f"\nCheckpoint saved: "
            f"{len(processed_videos)}/"
            f"{len(sampling_df)} videos"
        )


# =========================================================
# 6. Final save
# =========================================================

final_df = pd.DataFrame(results)

FINAL_METADATA = (
    "/kaggle/working/"
    "celebdf_aligned_faces_metadata.csv"
)

final_df.to_csv(
    FINAL_METADATA,
    index=False
)


print("\n===================================")
print("PROCESSING COMPLETED")
print("===================================")

print(
    "Total videos:",
    len(final_df)
)

print(
    "Total faces:",
    final_df["num_faces"].sum()
)

print(
    "Metadata:",
    FINAL_METADATA
)

print(
    "Aligned faces:",
    OUTPUT_ROOT
)
```
