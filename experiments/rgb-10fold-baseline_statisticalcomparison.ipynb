```python
# ============================================================
# STEP 7A — FRAME FILE HELPER
# ============================================================

def get_frame_files(video_dir):

    valid_extensions = (
        ".jpg",
        ".jpeg",
        ".png",
        ".bmp"
    )

    frame_files = [
        f for f in os.listdir(video_dir)
        if f.lower().endswith(valid_extensions)
    ]

    frame_files = sorted(frame_files)

    return frame_files


# Quick test
sample_video_dir = videos_df.iloc[0]["video_dir"]

print("Sample video:", sample_video_dir)

test_frames = get_frame_files(
    sample_video_dir
)

print("Number of frames:", len(test_frames))
print("First frames:", test_frames[:5])
```

```python
# ============================================================
# STEP 7 — CREATE RGB CACHE
# ============================================================

import os
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

RGB_CACHE_ROOT = "/kaggle/working/rgb_cache"
os.makedirs(RGB_CACHE_ROOT, exist_ok=True)

rgb_cache_records = []

for _, row in tqdm(
    videos_df.iterrows(),
    total=len(videos_df),
    desc="Creating RGB cache"
):
    
    video_id = row["video_id"]
    video_dir = row["video_dir"]
    label = row["label"]

    frame_files = get_frame_files(video_dir)

    for frame_file in frame_files:

        image_path = os.path.join(
            video_dir,
            frame_file
        )

        image = Image.open(
            image_path
        ).convert("RGB")

        image = image.resize(
            (224, 224),
            Image.Resampling.BILINEAR
        )

        image = np.asarray(
            image
        ).astype(np.uint8)

        cache_name = (
            f"{label}_{video_id}_{frame_file}"
            .replace("/", "_")
            .replace("\\", "_")
        )

        cache_path = os.path.join(
            RGB_CACHE_ROOT,
            cache_name + ".npy"
        )

        np.save(cache_path, image)

        rgb_cache_records.append({
            "video_id": video_id,
            "frame_file": frame_file,
            "label": label,
            "cache_path": cache_path
        })

rgb_cache_df = pd.DataFrame(
    rgb_cache_records
)

rgb_cache_df.to_csv(
    "/kaggle/working/rgb_cache_index.csv",
    index=False
)

print("\n✅ RGB CACHE COMPLETED")
print(
    "Total cached frames:",
    len(rgb_cache_df)
)
print(
    "Cache location:",
    RGB_CACHE_ROOT
)
```

```python
# ============================================================
# STEP 8 — FAST RGB DATASET
# ============================================================

import torch
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

class FastRGBDataset(Dataset):

    def __init__(
        self,
        dataframe,
        mode="train"
    ):

        self.mode = mode
        self.samples = []

        for video_id, group in dataframe.groupby(
            "video_id"
        ):

            group = group.sort_values(
                "frame_file"
            )

            frames = group.to_dict(
                "records"
            )

            if mode == "train":

                # First, middle, last
                n = len(frames)

                if n >= 3:
                    indices = [
                        0,
                        n // 2,
                        n - 1
                    ]

                    selected = [
                        frames[i]
                        for i in indices
                    ]

                else:
                    selected = frames

            else:

                # Maximum 5 validation frames
                selected = frames[:5]

            for sample in selected:
                self.samples.append(sample)

        print(
            f"{mode.upper()} samples:",
            len(self.samples)
        )

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):

        sample = self.samples[idx]

        # Load cached RGB image
        image = np.load(
            sample["cache_path"]
        )

        # HWC → CHW
        image = torch.from_numpy(
            image
        ).permute(
            2, 0, 1
        ).float() / 255.0

        # Already 224×224, but ensure size
        if image.shape[-2:] != (224, 224):

            image = F.interpolate(
                image.unsqueeze(0),
                size=(224, 224),
                mode="bilinear",
                align_corners=False
            ).squeeze(0)

        label = torch.tensor(
            sample["label"],
            dtype=torch.long
        )

        return (
            image,
            label,
            sample["video_id"]
        )
```

```python
# ============================================================
# STEP 12 — RGB EFFICIENTNET-B4
# ============================================================

import torch
import torch.nn as nn
from torchvision.models import (
    efficientnet_b4,
    EfficientNet_B4_Weights
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

weights = EfficientNet_B4_Weights.DEFAULT

model = efficientnet_b4(
    weights=weights
)

# Binary classification: Real / Fake
num_features = model.classifier[1].in_features

model.classifier[1] = nn.Linear(
    num_features,
    2
)

model = model.to(device)

# Same as DWT
if device.type == "cuda":
    model = model.to(
        memory_format=torch.channels_last
    )

print("EfficientNet-B4 ready")
print("Classifier:", model.classifier)
```

```python
# ============================================================
# STEP 13 — LOSS + OPTIMIZER + AMP
# ============================================================

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-5
)

scaler = torch.amp.GradScaler(
    "cuda",
    enabled=(device.type == "cuda")
)

print("Optimizer: Adam")
print("Learning rate: 1e-4")
print("Weight decay: 1e-5")
print(
    "AMP enabled:",
    device.type == "cuda"
)
```

```python
# ============================================================
# STEP 14 — FAST TRAINING FUNCTION
# ============================================================

def train_one_epoch_fast(
    model,
    loader,
    optimizer,
    criterion,
    scaler,
    device
):

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels, _ in loader:

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        # Same as DWT
        if device.type == "cuda":
            images = images.to(
                memory_format=torch.channels_last
            )

        optimizer.zero_grad(
            set_to_none=True
        )

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=(device.type == "cuda")
        ):

            logits = model(images)

            loss = criterion(
                logits,
                labels
            )

        scaler.scale(loss).backward()

        scaler.step(optimizer)

        scaler.update()

        running_loss += (
            loss.item() *
            labels.size(0)
        )

        predictions = torch.argmax(
            logits,
            dim=1
        )

        correct += (
            predictions == labels
        ).sum().item()

        total += labels.size(0)

    epoch_loss = (
        running_loss / total
    )

    epoch_accuracy = (
        correct / total
    )

    return epoch_loss, epoch_accuracy

print("✅ train_one_epoch_fast ready")
```

```python
# ============================================================
# STEP 15 — FAST VIDEO-LEVEL VALIDATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

import numpy as np


def validate_fast(
    model,
    loader,
    device
):

    model.eval()

    # video_id -> list of fake probabilities
    video_probs = {}

    # video_id -> true label
    video_labels = {}

    with torch.no_grad():

        for images, labels, video_ids in loader:

            images = images.to(
                device,
                non_blocking=True
            )

            if device.type == "cuda":
                images = images.to(
                    memory_format=torch.channels_last
                )

            with torch.autocast(
                device_type="cuda",
                dtype=torch.float16,
                enabled=(device.type == "cuda")
            ):

                logits = model(images)

            probabilities = torch.softmax(
                logits.float(),
                dim=1
            )

            fake_probs = probabilities[
                :, 1
            ].cpu().numpy()

            labels_np = labels.numpy()

            # Store frame probabilities by video
            for i, video_id in enumerate(video_ids):

                video_probs.setdefault(
                    video_id,
                    []
                ).append(
                    float(fake_probs[i])
                )

                video_labels[
                    video_id
                ] = int(labels_np[i])

    # ========================================================
    # VIDEO-LEVEL AGGREGATION
    # ========================================================

    y_true = []
    y_prob = []

    for video_id in video_probs:

        # Mean probability of validation frames
        mean_prob = np.mean(
            video_probs[video_id]
        )

        y_prob.append(mean_prob)

        y_true.append(
            video_labels[video_id]
        )

    y_true = np.array(y_true)
    y_prob = np.array(y_prob)

    y_pred = (
        y_prob >= 0.5
    ).astype(int)

    # ========================================================
    # METRICS
    # ========================================================

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    auc = roc_auc_score(
        y_true,
        y_prob
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "auc": auc,
        "y_true": y_true,
        "y_prob": y_prob,
        "y_pred": y_pred
    }


print("✅ validate_fast ready")
```

```python
# ============================================================
# STEP 16 — RGB FOLD 1 TRAINING
# ============================================================

import time

NUM_EPOCHS = 5

fold1_history = []
best_f1 = 0.0
best_fold1_results = None

for epoch in range(NUM_EPOCHS):

    start_time = time.time()

    # ----------------------------
    # Training
    # ----------------------------
    train_loss, train_acc = train_one_epoch_fast(
        model,
        train_loader,
        optimizer,
        criterion,
        scaler,
        device
    )

    # ----------------------------
    # Validation
    # ----------------------------
    val_results = validate_fast(
        model,
        val_loader,
        device
    )

    epoch_time = time.time() - start_time

    print(
        f"\nEpoch {epoch + 1}/{NUM_EPOCHS}"
    )

    print(
        f"Train Loss: {train_loss:.4f} | "
        f"Train Accuracy: {train_acc:.4f}"
    )

    print(
        f"Val Accuracy: "
        f"{val_results['accuracy']:.4f}"
    )

    print(
        f"Precision: "
        f"{val_results['precision']:.4f} | "
        f"Recall: "
        f"{val_results['recall']:.4f}"
    )

    print(
        f"F1: "
        f"{val_results['f1']:.4f} | "
        f"AUC: "
        f"{val_results['auc']:.4f}"
    )

    print(
        f"Epoch Time: "
        f"{epoch_time / 60:.2f} min"
    )

    # ----------------------------
    # Save history
    # ----------------------------
    fold1_history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "val_accuracy": val_results["accuracy"],
        "precision": val_results["precision"],
        "recall": val_results["recall"],
        "f1": val_results["f1"],
        "auc": val_results["auc"],
        "epoch_time_min": epoch_time / 60
    })

    # ----------------------------
    # Best F1 model
    # ----------------------------
    if val_results["f1"] > best_f1:

        best_f1 = val_results["f1"]

        best_fold1_results = {
            "fold": 1,
            "accuracy": val_results["accuracy"],
            "precision": val_results["precision"],
            "recall": val_results["recall"],
            "f1": val_results["f1"],
            "auc": val_results["auc"],
            "y_true": val_results["y_true"],
            "y_prob": val_results["y_prob"],
            "y_pred": val_results["y_pred"]
        }

        torch.save(
            {
                "fold": 1,
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "metrics": {
                    "accuracy": val_results["accuracy"],
                    "precision": val_results["precision"],
                    "recall": val_results["recall"],
                    "f1": val_results["f1"],
                    "auc": val_results["auc"]
                }
            },
            "/kaggle/working/fold_1_best_rgb.pth"
        )

        print("🔥 Fold-1 best RGB model saved!")
```

```python
# ============================================================
# STEP 17 — FOLDS 2 TO 10
# ============================================================

import os
import time
import pandas as pd

RESULT_DIR = "/kaggle/working/10fold_rgb_results"
os.makedirs(RESULT_DIR, exist_ok=True)


# ------------------------------------------------------------
# Fresh EfficientNet-B4 for every fold
# ------------------------------------------------------------
def create_fresh_rgb_model():

    weights = EfficientNet_B4_Weights.DEFAULT

    new_model = efficientnet_b4(
        weights=weights
    )

    num_features = (
        new_model.classifier[1].in_features
    )

    new_model.classifier[1] = nn.Linear(
        num_features,
        2
    )

    new_model = new_model.to(device)

    if device.type == "cuda":
        new_model = new_model.to(
            memory_format=torch.channels_last
        )

    return new_model


# ------------------------------------------------------------
# Store Fold-1 result
# ------------------------------------------------------------

all_fold_results = []

all_fold_results.append({
    "fold": 1,
    "accuracy": best_fold1_results["accuracy"],
    "precision": best_fold1_results["precision"],
    "recall": best_fold1_results["recall"],
    "f1": best_fold1_results["f1"],
    "auc": best_fold1_results["auc"]
})


# ------------------------------------------------------------
# FOLDS 2 → 10
# ------------------------------------------------------------

for fold in range(2, 11):

    print("\n" + "=" * 70)
    print(f"🚀 STARTING RGB FOLD {fold}/10")
    print("=" * 70)

    # --------------------------------------------------------
    # VIDEO-LEVEL SPLIT
    # --------------------------------------------------------

    train_videos = videos_df[
        videos_df["fold"] != fold
    ].copy()

    val_videos = videos_df[
        videos_df["fold"] == fold
    ].copy()

    # --------------------------------------------------------
    # RGB CACHE SPLIT
    # --------------------------------------------------------

    train_rgb_df = rgb_cache_df[
        rgb_cache_df["video_id"].isin(
            train_videos["video_id"]
        )
    ].copy()

    val_rgb_df = rgb_cache_df[
        rgb_cache_df["video_id"].isin(
            val_videos["video_id"]
        )
    ].copy()

    # --------------------------------------------------------
    # DATASETS
    # --------------------------------------------------------

    train_dataset = FastRGBDataset(
        train_rgb_df,
        mode="train"
    )

    val_dataset = FastRGBDataset(
        val_rgb_df,
        mode="val"
    )

    # --------------------------------------------------------
    # DATALOADERS
    # --------------------------------------------------------

    train_loader = DataLoader(
        train_dataset,
        batch_size=32,
        shuffle=True,
        num_workers=4,
        pin_memory=True,
        persistent_workers=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=32,
        shuffle=False,
        num_workers=4,
        pin_memory=True,
        persistent_workers=True
    )

    print(
        "Train samples:",
        len(train_dataset)
    )

    print(
        "Validation samples:",
        len(val_dataset)
    )

    # --------------------------------------------------------
    # FRESH MODEL
    # --------------------------------------------------------

    model = create_fresh_rgb_model()

    # --------------------------------------------------------
    # FRESH LOSS + OPTIMIZER
    # --------------------------------------------------------

    criterion = nn.CrossEntropyLoss()

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=1e-4,
        weight_decay=1e-5
    )

    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=(device.type == "cuda")
    )

    # --------------------------------------------------------
    # TRAINING
    # --------------------------------------------------------

    NUM_EPOCHS = 5

    best_f1 = 0.0
    best_result = None
    best_predictions = None

    for epoch in range(NUM_EPOCHS):

        epoch_start = time.time()

        train_loss, train_acc = (
            train_one_epoch_fast(
                model,
                train_loader,
                optimizer,
                criterion,
                scaler,
                device
            )
        )

        val_results = validate_fast(
            model,
            val_loader,
            device
        )

        epoch_time = (
            time.time() - epoch_start
        )

        print(
            f"\nFold {fold} | "
            f"Epoch {epoch + 1}/{NUM_EPOCHS}"
        )

        print(
            f"Train Loss: {train_loss:.4f} | "
            f"Train Acc: {train_acc:.4f}"
        )

        print(
            f"Val Acc: "
            f"{val_results['accuracy']:.4f} | "
            f"Precision: "
            f"{val_results['precision']:.4f}"
        )

        print(
            f"Recall: "
            f"{val_results['recall']:.4f} | "
            f"F1: "
            f"{val_results['f1']:.4f} | "
            f"AUC: "
            f"{val_results['auc']:.4f}"
        )

        print(
            f"Time: "
            f"{epoch_time / 60:.2f} min"
        )

        # ----------------------------------------------------
        # BEST F1
        # ----------------------------------------------------

        if val_results["f1"] > best_f1:

            best_f1 = val_results["f1"]

            best_result = {
                "fold": fold,
                "accuracy": val_results["accuracy"],
                "precision": val_results["precision"],
                "recall": val_results["recall"],
                "f1": val_results["f1"],
                "auc": val_results["auc"]
            }

            # Save predictions for ROC/CM later
            best_predictions = {
                "y_true": val_results["y_true"],
                "y_prob": val_results["y_prob"],
                "y_pred": val_results["y_pred"]
            }

            # ------------------------------------------------
            # SAVE BEST MODEL
            # ------------------------------------------------

            torch.save(
                {
                    "fold": fold,
                    "epoch": epoch + 1,
                    "model_state_dict":
                        model.state_dict(),
                    "optimizer_state_dict":
                        optimizer.state_dict(),
                    "metrics": best_result
                },
                os.path.join(
                    RESULT_DIR,
                    f"fold_{fold}_best_rgb.pth"
                )
            )

            print(
                "🔥 Best RGB model saved!"
            )

    # --------------------------------------------------------
    # SAVE FOLD RESULT
    # --------------------------------------------------------

    all_fold_results.append(
        best_result
    )

    # --------------------------------------------------------
    # SAVE FOLD PREDICTIONS
    # --------------------------------------------------------

    np.savez(
        os.path.join(
            RESULT_DIR,
            f"fold_{fold}_predictions.npz"
        ),
        y_true=best_predictions["y_true"],
        y_prob=best_predictions["y_prob"],
        y_pred=best_predictions["y_pred"]
    )

    # --------------------------------------------------------
    # SAVE PROGRESS
    # --------------------------------------------------------

    pd.DataFrame(
        all_fold_results
    ).to_csv(
        os.path.join(
            RESULT_DIR,
            "rgb_10fold_results_progress.csv"
        ),
        index=False
    )

    print(
        f"\n✅ RGB Fold {fold} completed."
    )
    print(
        f"Best F1: {best_result['f1']:.4f}"
    )
```

```python
# ============================================================
# DWT vs RGB — FINAL STATISTICAL COMPARISON
# ============================================================

import os
import numpy as np
import pandas as pd
from scipy.stats import ttest_rel, t

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

DWT_PATH = (
    "/kaggle/input/datasets/"
    "srujanamannepuli/"
    "papaer-results/"
    "CSV_Files/"
    "DWT_10Fold_Foldwise_Results.csv"
)

RGB_PATH = (
    "/kaggle/working/"
    "10fold_rgb_results/"
    "rgb_10fold_results_progress.csv"
)

STAT_DIR = (
    "/kaggle/working/"
    "DWT_vs_RGB_Statistical_Analysis"
)

os.makedirs(
    STAT_DIR,
    exist_ok=True
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

dwt = pd.read_csv(DWT_PATH)
rgb = pd.read_csv(RGB_PATH)

# Standardize column names
dwt.columns = [
    str(c).strip().lower()
    for c in dwt.columns
]

rgb.columns = [
    str(c).strip().lower()
    for c in rgb.columns
]

print("=" * 70)
print("DWT vs RGB")
print("=" * 70)

print("\nDWT:")
print(dwt.columns.tolist())

print("\nRGB:")
print(rgb.columns.tolist())

# ------------------------------------------------------------
# CHECK FOLDS
# ------------------------------------------------------------

print("\nDWT folds:")
print(dwt["fold"].tolist())

print("\nRGB folds:")
print(rgb["fold"].tolist())

# ------------------------------------------------------------
# KEEP REQUIRED COLUMNS
# ------------------------------------------------------------

metrics = [
    "accuracy",
    "precision",
    "recall",
    "f1",
    "auc"
]

# ------------------------------------------------------------
# MERGE SAME FOLDS
# ------------------------------------------------------------

comparison = pd.merge(
    dwt[
        ["fold"] +
        metrics
    ],
    rgb[
        ["fold"] +
        metrics
    ],
    on="fold",
    how="inner",
    suffixes=(
        "_dwt",
        "_rgb"
    )
)

comparison = comparison.sort_values(
    "fold"
).reset_index(drop=True)

print(
    "\nMatched folds:",
    len(comparison)
)

display(comparison)

# ------------------------------------------------------------
# SAVE FOLD-WISE COMPARISON
# ------------------------------------------------------------

comparison_path = os.path.join(
    STAT_DIR,
    "DWT_vs_RGB_Foldwise_Comparison.csv"
)

comparison.to_csv(
    comparison_path,
    index=False
)

# ============================================================
# STATISTICAL TESTS
# ============================================================

results = []

for metric in metrics:

    dwt_values = comparison[
        f"{metric}_dwt"
    ].astype(float).values

    rgb_values = comparison[
        f"{metric}_rgb"
    ].astype(float).values

    # --------------------------------------------------------
    # Means and SDs
    # --------------------------------------------------------

    dwt_mean = np.mean(
        dwt_values
    )

    dwt_sd = np.std(
        dwt_values,
        ddof=1
    )

    rgb_mean = np.mean(
        rgb_values
    )

    rgb_sd = np.std(
        rgb_values,
        ddof=1
    )

    # --------------------------------------------------------
    # Paired difference
    # --------------------------------------------------------

    difference = (
        dwt_values -
        rgb_values
    )

    mean_difference = np.mean(
        difference
    )

    difference_sd = np.std(
        difference,
        ddof=1
    )

    # --------------------------------------------------------
    # Paired t-test
    # --------------------------------------------------------

    t_statistic, p_value = ttest_rel(
        dwt_values,
        rgb_values
    )

    # --------------------------------------------------------
    # 95% CI
    # --------------------------------------------------------

    n = len(difference)

    standard_error = (
        difference_sd /
        np.sqrt(n)
    )

    t_critical = t.ppf(
        0.975,
        n - 1
    )

    ci_lower = (
        mean_difference -
        t_critical *
        standard_error
    )

    ci_upper = (
        mean_difference +
        t_critical *
        standard_error
    )

    # --------------------------------------------------------
    # Cohen's d for paired samples
    # --------------------------------------------------------

    if difference_sd > 0:

        cohens_d = (
            mean_difference /
            difference_sd
        )

    else:

        cohens_d = np.nan

    # --------------------------------------------------------
    # Significance
    # --------------------------------------------------------

    significance = (
        "Significant"
        if p_value < 0.05
        else "Not Significant"
    )

    results.append({

        "Metric":
            metric.upper(),

        "DWT Mean":
            dwt_mean,

        "DWT SD":
            dwt_sd,

        "RGB Mean":
            rgb_mean,

        "RGB SD":
            rgb_sd,

        "Mean Difference (DWT-RGB)":
            mean_difference,

        "95% CI Lower":
            ci_lower,

        "95% CI Upper":
            ci_upper,

        "t-statistic":
            t_statistic,

        "p-value":
            p_value,

        "Cohen's d":
            cohens_d,

        "Significance":
            significance
    })


stats_df = pd.DataFrame(
    results
)

# ============================================================
# BONFERRONI CORRECTION
# ============================================================

num_tests = len(
    stats_df
)

stats_df[
    "Bonferroni p-value"
] = np.minimum(
    stats_df["p-value"] * num_tests,
    1.0
)

stats_df[
    "Bonferroni Significance"
] = np.where(
    stats_df["Bonferroni p-value"] < 0.05,
    "Significant",
    "Not Significant"
)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

stats_path = os.path.join(
    STAT_DIR,
    "DWT_vs_RGB_Statistical_Comparison.csv"
)

stats_df.to_csv(
    stats_path,
    index=False
)

# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL STATISTICAL RESULTS")
print("=" * 70)

display(
    stats_df.round(6)
)

print("\n✅ Statistical comparison completed!")

print(
    "\nFold-wise file:",
    comparison_path
)

print(
    "Statistical file:",
    stats_path
)
```
